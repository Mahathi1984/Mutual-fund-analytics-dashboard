```python
import pandas as pd
import numpy as np

# NAV Dataset
dates = pd.date_range('2025-01-01', '2025-12-31')
nav_data = []

for code in [120503,120504,120505]:
    nav = 50
    for d in dates:
        nav += np.random.uniform(-1,1)
        nav_data.append([code,d.strftime('%Y-%m-%d'),round(nav,2)])

nav_df = pd.DataFrame(nav_data, columns=['amfi_code','date','nav'])
nav_df.to_csv('nav_history.csv', index=False)

# Investor Dataset
transactions = []
for i in range(1,501):
    transactions.append([
        f'INV{i:03}',
        np.random.choice([120503,120504,120505]),
        pd.Timestamp('2025-01-01') + pd.Timedelta(days=np.random.randint(0,365)),
        np.random.choice(['SIP','Lumpsum','Redemption']),
        np.random.randint(1000,50000),
        round(np.random.uniform(10,500),2),
        np.random.choice(['Verified','Pending'])
    ])

txn_df = pd.DataFrame(transactions,
columns=['investor_id','scheme_code','transaction_date',
         'transaction_type','amount','units','kyc_status'])

txn_df.to_csv('investor_transactions.csv', index=False)

print("Files created successfully")
```
